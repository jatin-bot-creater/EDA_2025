**Pie Chart Automate**

```python
for i in cat[1:]:
    continent_key=visa_df[i].value_counts().keys()
    continent_values=visa_df[i].value_counts().values
    #plt.pie(continent_key,continent_values)
    plt.pie(continent_values,
           labels=continent_key,
           autopct="%0.2f%%",
           startangle=45)
    
    # Step-2:
    folder_name='Pie_chart'
    os.makedirs(folder_name,exist_ok=True)

    # Step-3:
    cwd=os.getcwd()
    new_dir=os.path.join(cwd,folder_name)
    file_name=f'{i}.jpg'
    path=os.path.join(new_dir,file_name)

    # Step-4:
    plt.savefig(path)
    plt.show()
```
